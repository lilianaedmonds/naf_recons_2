```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("/home/lilianae/projects/naf_clean/admm_results_ismrm/Subject3_MID0083/hwhm_output/UPDATE_interface_43_160_128/metrics_summary.csv")

plt.figure(figsize=(6, 4))

sns.boxplot(data=df, x="method", y="hwhm")

plt.title("HWHM Comparison (Subject)")
plt.ylabel("HWHM (mm)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="dataset", y="hwhm", hue="method")

plt.title("HWHM Across Datasets")
plt.ylabel("HWHM (mm)")
plt.grid(alpha=0.3)

plt.legend(title="Method")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# 1. Boxplot handles the color palette
sns.boxplot(data=df, x="dataset", y="hwhm", hue="method", palette="Set2")

# 2. Stripplot uses the hue for 'dodging' (alignment) 
# but we explicitly set the palette to a list of blacks
# This satisfies the new Seaborn requirement and removes the warning
sns.stripplot(
    data=df, 
    x="dataset", 
    y="hwhm", 
    hue="method", 
    dodge=True, 
    palette=["black"] * df["method"].nunique(), # Creates a list of blacks for each hue level
    alpha=0.5,
    legend=False
)

plt.title("HWHM Across Datasets (with individual points)")
plt.ylabel("HWHM (mm)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Barplot will show the colors clearly even with single values
sns.barplot(data=df, x="dataset", y="hwhm", hue="method", palette="Set2", edgecolor="black")

# Keep the dots if you like the aesthetic
sns.stripplot(data=df, x="dataset", y="hwhm", hue="method", 
              dodge=True, color="black", alpha=0.5, legend=False)

plt.title("HWHM Across Datasets - Heart Interface")
plt.ylabel("HWHM (mm)")
plt.show()

plt.figure(figsize=(8, 5))

# Barplot will show the colors clearly even with single values
sns.barplot(data=df, x="dataset", y="sharpness", hue="method", palette="Set2", edgecolor="black")

# Keep the dots if you like the aesthetic
sns.stripplot(data=df, x="dataset", y="sharpness", hue="method", 
              dodge=True, color="black", alpha=0.5, legend=False)

plt.title("Image Sharpness Across Datasets - Heart Interface")
plt.ylabel("1/HWHM (mm)")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="dataset", y="hwhm", hue="method")
sns.stripplot(data=df, x="dataset", y="hwhm", hue="method",
              dodge=True, color="black", alpha=0.5)

plt.title("HWHM Across Datasets (with individual points)")
plt.ylabel("HWHM (mm)")
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# 1. Keep the boxplot but make it thin/grey so it acts as a marker line
sns.boxplot(data=df, x="dataset", y="hwhm", hue="method", 
            palette="Set2", showfliers=False, 
            linewidth=1.5, boxprops=dict(alpha=0.3))

# 2. Color the dots based on 'method'
# This is where your visual 'key' will come from
sns.stripplot(data=df, x="dataset", y="hwhm", hue="method", 
              dodge=True, palette="Set2", size=8, jitter=False)

# 3. Clean up the legend (prevents duplicate entries)
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles[0:2], labels[0:2], title="Method")

plt.title("HWHM Across Datasets")
plt.ylabel("HWHM (mm)")
plt.grid(alpha=0.3)
plt.show()
```
